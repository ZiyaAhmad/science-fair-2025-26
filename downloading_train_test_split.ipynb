```python
import numpy as np
from sklearn.model_selection import train_test_split
from google.colab import drive

drive.mount('/content/drive')

# change this to wherever your ecg_data.npz is saved!
save_path = '/content/drive/MyDrive/ecg_data.npz'
data = np.load(save_path)
all_windows = data['windows']
all_labels = data['labels']

print("full dataset distribution:")
print(f"  normal (0): {np.sum(all_labels == 0):,}")
print(f"  sva (1): {np.sum(all_labels == 1):,}")
print(f"  pre-sva (2): {np.sum(all_labels == 2):,}")
print(f"  total: {len(all_labels):,}")

# create train/test split using 3-class stratification
train_indices, test_indices = train_test_split(
    np.arange(len(all_labels)),  # just split indices
    test_size=0.2,
    random_state=42,
    stratify=all_labels  # stratify on 3-class labels
)

print(f"\ntrain/test split created:")
print(f"  train indices: {len(train_indices):,}")
print(f"  test indices: {len(test_indices):,}")

# verify test set distribution
test_labels = all_labels[test_indices]
print(f"\ntest set distribution:")
print(f"  normal (0): {np.sum(test_labels == 0):,}")
print(f"  sva (1): {np.sum(test_labels == 1):,}")
print(f"  pre-sva (2): {np.sum(test_labels == 2):,}")

# save the indices (change the path below)
split_path = '/content/drive/MyDrive/train_test_split.npz'
np.savez(split_path,
         train_indices=train_indices,
         test_indices=test_indices)

print(f"\n train/test split saved to:")
print(f"   {split_path}")
```
